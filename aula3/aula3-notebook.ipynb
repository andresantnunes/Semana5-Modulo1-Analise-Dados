```python
import pandas as pd

df = pd.read_csv('estoque_produtos_100_itens.csv',
                sep=";"
            )

print()
print("df.head(3)")
print(df.head(3))
```

```python
# Filtrando pela string, se contem ou não um item
df_produtos_gourmet = df[(df['Descricao'].str.contains('A'))]

display(df_produtos_gourmet)

# termina com uma string
df_produtos_gourmet = df[(df['Descricao'].str.endswith('Especial'))]

display(df_produtos_gourmet)

# começa com uma string
df_produtos_gourmet = df[(df['Descricao'].str.startswith('Achocolatado '))]

display(df_produtos_gourmet)
```

```python
# regex de pesquisa de multiplos itens
filtro = r"Especial|Gourmet"

df_produtos_gourmet = df[(df['Descricao'].str.contains(filtro))]

display(df_produtos_gourmet)
display(df_produtos_gourmet['Preco'].mean())
```

```python
# Criando um DataFrame de exemplo
dados = {
    'Nome': 'Carlos Ana Bruno Diego Helena'.split(),
    'Email': ['carlos@gmail.com', 'ana@outlook.com', 'bruno@yahoo.com', 'diego@gmail.com', 'helena@empresa.com.br']
}
df = pd.DataFrame(dados)

# Regex para buscar 'gmail' OU 'yahoo'
padrao = r'(gmail|yahoo)'

# Filtrando o DataFrame
resultado = df[df['Email'].str.contains(padrao, case=False, na=False)]
print(resultado)
```

```python
# Regex com dois grupos de captura: (usuario) @ (dominio)
padrao_extrair = r'(?P<Usuario>[\w\.-]+)@(?P<Dominio>[\w\.-]+)'

# Dá pra fazer mas não é usual
# O mais comum é mapear de um dataframe para outro
# Não é comum usar regex para fazer dataframes

# Extraindo para novas colunas
df_extraido = df['Email'].str.extract(padrao_extrair)
print(df_extraido)
```

```python
df_produtos = pd.DataFrame({'Preço': ['R$ 1.500,00', 'R$ 350,50', 'R$ 89,90']})

# Queremos buscar tudo o que NÃO for número e ponto/vírgula para remover
# \D no regex significa "tudo que não é dígito"

# Substitui tudo que não é digito por nada
# O dataframe aplica operações do str em todos os itens da coluna Preço
# todos os itens irão executar o replace(item_a_ser_substituido, item_a_substituir, regex=True ou False)

#Estamos salvando os valores atualizado da coluna Preço na coluna Preço_Limpo
df_produtos['Preço_Limpo'] = df_produtos['Preço'].str.replace(r'[^\d,]', '', regex=True)
display(df_produtos)
```
