```python
import pandas as pd
import numpy as np

df  = pd.DataFrame({
	'produto': ['Notebook', 'Mouse','Monitor' ,  'teclado'],
	'preco': [4500.1111, 150.2222, 1500.2222,320.2222313]
})
df['preco_ajustado'] = np.round(df['preco'] * 1.10, 2) # arredondamento para 2 casas decimais
display(df['preco_ajustado'])

# Operação acontece para todos os itens de preço sem a necessidade de loop
df['preco_ajustado'] = (df['preco'] - np.mean(df['preco'])) - np.std(df['preco']) #escalonamento de dados

# Normalização são usadas para graficos e para estatisticas
df['preco_normalizado_z'] = (df['preco'] - np.mean(df['preco'])) / np.std(df['preco'])

print(df)
print(df['preco_normalizado_z'].mean())
print(df['preco_normalizado_z'].std())

```
