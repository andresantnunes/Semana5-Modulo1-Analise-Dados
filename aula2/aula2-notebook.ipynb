```python
df = df_clientes
df.to_json("cliente.json",orient='split')
# Output: {"index": [...], "columns": [...], "data": [[...], [...]]}

# formato mais comum de json
df.to_json("cliente1.json",orient='records')
# Output: [{"col1": value1, "col2": value2}, {...}]

df.to_json("cliente2.json",orient='index')
# Output: {"row1": {"col1": value1, "col2": value2}, ...}

# cada coluna vira uma serie
df.to_json("cliente3.json",orient='columns')
# Output: {"col1": {"row1": value1, "row2": value2}, ...}

df.to_json("cliente4.json",orient='values')
# Output: [[value1, value2], [...]]

df.to_json("cliente5.json",orient='table')
# Output: {"schema": {...}, "data": [{"index": ..., "col1": ..., ...}]}
```

```python
import matplotlib.pyplot as plt

# Criando o gráfico de caixa (Boxplot) para ver os extremos
df_produtos[['Estoque', 'Preco']].plot(kind='box', subplots=True, layout=(1,2), figsize=(10,5))
plt.show()

# Criando o histograma para ver a frequência
df_produtos['Preco'].hist(bins=5)
plt.title('Distribuição de Preços')
plt.show()
```

```python
# Filtrando os produtos para terem estoque menor que 200 e preco menor que 30
df_produtos_filtrados = df_produtos[(df_produtos['Estoque'] < 200) &
                                    (df_produtos['Preco'] < 30)]

# Criando o gráfico de caixa (Boxplot) para ver os extremos
df_produtos_filtrados[['Estoque', 'Preco']].plot(kind='box', subplots=True, layout=(1,2), figsize=(10,5))
plt.show()

# Criando o histograma para ver a frequência
df_produtos_filtrados['Preco'].hist(bins=5)
plt.title('Distribuição de Preços')
plt.show()
```

```python
# Filtrando com operação ou
df_produtos_filtrados = df_produtos[(df_produtos['Estoque'] > 100) |
                                    (df_produtos['Preco'] > 20)]

display(df_produtos_filtrados)

```
